```python
# Functions
def compute_sum(values):
    total = 0.0
    for v in values:
        total += v
    return total
print(compute_sum([1.0, 2.0, 3.0]))      # 6.0

def mean_of(*values):
    total = 0.0
    for v in values:
        total += v
    return total / len(values)

print(round(mean_of(6.1, 7.4, 5.9), 2))  # 6.47
```
