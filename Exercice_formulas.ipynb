````python
# 1. Lists
temps = [-2.3, -1.1, 0.4, 1.2, -0.8]
temps.append(-3.1)             # add at the end
temps.remove(0.4)              # remove by VALUE (first match)

print(temps[0])                # first            -> -2.3
print(temps[-1])               # last             -> -3.1
print(temps[1:3])              # 2nd and 3rd      -> [-1.1, 1.2]   (stop is excluded)
print(temps[:4])               # prints the 0, 1, 2, 3
print(len(temps))              # number of items  -> 5
#Slice cheat: `x[start:stop:step]

# 2. loope for lists
planets = ["mercury", "venus", "earth", "mars", "jupiter", "saturn", "uranus", "neptune"]
for planet in planets:         # keep names ending in "s"
    if planet[-1] != "s":
        continue               # skip the others explicitly
    print(planet)              # venus, mars, uranus

dates = ["Mon", "Tue", "Wed", "Thu", "Fri"]
jfj_celsius = [-2.3, -1.1, 0.4, 1.2, -0.8]
bas_celsius = [6.1, 7.4, 5.9, 8.2, 6.8]

for date, jfj, bas in zip(dates, jfj_celsius, bas_celsius):
    print(f"{date}: {bas - jfj:.2f}")          # Mon: 8.40 ... Fri: 7.60

week_celsius = [6.1, 7.4, 5.9, 8.2, 11.3, 9.8, 6.8]

warmest_temp = None                    # define BEFORE the loop
warmest_day = None
for i, temp in enumerate(week_celsius):
    day = i + 1                        # enumerate starts at 0 (or: enumerate(x, start=1))
    print(day, temp)
    if warmest_temp is None or temp > warmest_temp:
        warmest_temp = temp
        warmest_day = day
print(warmest_temp, warmest_day)       # 11.3 5

mixed_celsius = [6.1, None, 5.9, None, 9.5, 6.8]
first_hot = None                       # defined before: safe if nothing exceeds 9.0
i = 0
while i < len(mixed_celsius):
    value = mixed_celsius[i]
    i += 1                             # update the index BEFORE any continue, or infinite loop
    if value is None:
        continue                       # jump back to the while line
    if value > 9.0:
        first_hot = value
        break                          # leave the loop entirely
print(first_hot)                       # 9.5

## 3. Dictionaries

station_records = [("JFJ", "Jungfraujoch", 3571),
                   ("BAS", "Basel-Binningen", 316),
                   ("LUG", "Lugano", 273)]
elevations = {}
for code, name, elevation_m in station_records:   # unpack each tuple
    elevations[code] = elevation_m                # d[key] = value
print(elevations["JFJ"])                  # 3571

del elevations["LUG"]                     # delete an entry
for code, elevation_m in elevations.items():   # key and value together
    print(code, elevation_m)

# Ex 17 Q5-Q8: planet masses, ALL in units of 1e24 kg
planet_masses_1e24kg = {
    "mercury": 0.330,
    "venus": 4.87,
    "earth": 5.97,
    "mars": 0.642,
    "jupiter": 1898,
    "saturn": 568,
    "uranus": 86.8,
    "neptune": 102,
}
print(planet_masses_1e24kg["earth"])      # 5.97

heavy = []                                # planets heavier than 100e24 kg
for name, mass in planet_masses_1e24kg.items():
    if mass > 100:
        heavy.append(name)
print(heavy)                              # ['jupiter', 'saturn', 'neptune']

planet_masses_1e24kg["pluto"] = 0.0130    # add: one assignment
for name, mass in planet_masses_1e24kg.items():
    print(name)

months = ["01", "02", "01", "01", "02"]
counts = {}
for month in months:
    counts[month] = counts.get(month, 0) + 1    # .get(key, 0) = 0 the first time
print(counts)                                   # {'01': 3, '02': 2}
# same idea for sums: totals[key] = totals.get(key, 0) + value

# 4. Tuples
station_coords = (46.5475, 7.9853)
lat, lon = station_coords                # unpacking: one line, two variables
print(lat, lon)

## 5. Functions  (Ex 3, 4, 8, 9, 10, 11, 14)
def mean(values):
    return sum(values) / len(values)

print(round(mean([6.1, 7.4, 5.9, 8.2, 6.8]), 2))   # 6.88
print(mean.__doc__)                                # the docstring text
help(mean)                                         # what a reader sees

def is_warm(temp_celsius, threshold=10.0):
    if temp_celsius >= threshold:
        return True
    else:
        return False
print(is_warm(6.1))                  # False  (default threshold 10.0)
print(is_warm(6.1, threshold=5.0))   # True   (keyword argument)

def summarise(values, scale=1.0):
    scaled = [v * scale for v in values]
    return sum(scaled), sum(scaled) / len(scaled)      # returns a tuple

readings = [6.1, 7.4, 5.9, 8.2, 6.8]
total_celsius, mean_celsius = summarise(readings)      # unpack into two names
_, mean_doubled = summarise(readings, scale=2.0)       # _ = value we ignore
print(round(mean_doubled, 2))                          # 13.76
result = summarise(readings)                           # one name -> the whole tuple
print(type(result))                                    # <class 'tuple'>
```

**Variable number of arguments** `*args` (Ex 9):

```python
def warmest(*temps_celsius):           # temps_celsius is a tuple of all positional args
    best = None
    for t in temps_celsius:            # same running-max pattern as in §4
        if best is None or t > best:
            best = t
    return best

print(warmest(6.1, 7.4, 5.9))          # 7.4
print(warmest(*[6.1, 7.4, 5.9, 8.2, 6.8]))   # 8.2  (* unpacks the list into arguments)
```

**Mutable default bug** (Ex 10) — and **pure functions** (Ex 14):

```python
# BUG: def collect_above(values, threshold, out=[])
# The default list is created ONCE, when the function is defined, and shared by every call,
# so the second call keeps appending to the result of the first.

def collect_above(values, threshold, out=None):    # FIX: None as default...
    if out is None:
        out = []                                   # ...and a fresh list on each call
    for v in values:
        if v > threshold:
            out.append(v)
    return out

first = collect_above([1, 5, 9], 4)
second = collect_above([2, 6, 10], 4)
print(first, second)                               # [5, 9] [6, 10]
```

```python
def drop_last_impure(readings):        # IMPURE: pop() changes the caller's list
    readings.pop()
    return readings

original = [6.1, 7.4, 5.9]
drop_last_impure(original)
print(original)                        # [6.1, 7.4]  <- the caller's list was modified

def drop_last(readings):               # PURE: the slice builds a NEW list
    return readings[:-1]

original = [6.1, 7.4, 5.9]
print(drop_last(original), original)   # [6.1, 7.4] [6.1, 7.4, 5.9]  original unchanged

# Link to the mutable-default bug: in both cases a function silently changes a list that
# outlives the call (the shared default, or the caller's list). Pure = build and return new data.
```

---

## 6. Comprehensions  (Ex 13, Ex 2)

```python
# dict comprehension  {key: value for ... in ...}
elevations_by_code = {code: elev for code, name, elev in station_records}

stations = {
    "JFJ": {"readings_celsius": [-2.3, -1.1, 0.4, 1.2, -0.8]},
    "BAS": {"readings_celsius": [6.1, 7.4, 5.9, 8.2, 6.8]},
    "LUG": {"readings_celsius": [9.4, 10.2, 8.8, 11.1, 9.9]},
}

means = {code: mean(data["readings_celsius"]) for code, data in stations.items()}
print({code: round(m, 2) for code, m in means.items()})   # {'JFJ': -0.52, 'BAS': 6.88, 'LUG': 9.88}

# list comprehension  [expr for ... in ... if condition]
below_zero = [code for code, m in means.items() if m < 0]
print(below_zero)                                         # ['JFJ']
```

---

## 7. Errors: `try` / `except` / `raise`  (Ex 15, also Ex 12 and 16 step 7)

```python
def to_kelvin(temp_celsius: float) -> float:
    """Return the temperature in kelvin; raise ValueError below -273.15 °C (absolute zero)."""
    if temp_celsius < -273.15:
        raise ValueError(f"{temp_celsius} °C is below absolute zero")
    return temp_celsius + 273.15

for temp in [20.0, -300.0, 0.0]:
    try:
        print(temp, "->", to_kelvin(temp))
    except ValueError as error:          # the loop keeps going after the error
        print("invalid:", error)
```

Common exceptions: `ValueError` (bad value, e.g. `float("abc")`, `float("")`), `TypeError` (e.g. item assignment on a tuple), `KeyError` (missing dict key), `IndexError`, `ZeroDivisionError`.

---

## 8. A year of station data  (Ex 16, steps 1–7)

Pre-supplied cell — downloads the file and defines `data_file`:

```python
import pooch

data_file = pooch.retrieve(
    url="https://raw.githubusercontent.com/gse-unil/2026_MLEES_book/main/data/part-I/station_iib_daily_max_temp_2022.csv",
    known_hash="sha256:034755fb289b4e157e5d029995481786a359eb25a80df62c09ee83d063013144",
    fname="station_iib_daily_max_temp_2022.csv",
    path=pooch.os_cache("mlees"),
)
```

**Step 1 — conversion function**

```python
def fahrenheit_to_celsius(temp_fahrenheit):
    """Return the temperature in °C for an input in °F."""
    return (temp_fahrenheit - 32) * 5 / 9

print(fahrenheit_to_celsius(32))     # 0.0
print(fahrenheit_to_celsius(212))    # 100.0
```

**Steps 2, 3, 4 — one loop does all three** (read, skip gaps and count, per-month counts and sums, then means):

```python
from pathlib import Path

with Path(data_file).open("r", encoding="utf-8") as f:   # read
    lines = f.readlines()                                # list of strings

n_converted = 0
n_skipped = 0
counts = {}                                  # month -> number of observations
totals_c = {}                                # month -> running sum in °C

for line in lines[1:]:                       # [1:] skips the header
    fields = line.strip().split(",")         # "IIB,01.01.22,30.2" -> ["IIB", "01.01.22", "30.2"]
    if fields[2] == "":                      # empty reading -> skip and count
        n_skipped += 1
        continue
    month = fields[1].split(".")[1]          # "01.01.22" -> "01"
    temp_c = fahrenheit_to_celsius(float(fields[2]))
    n_converted += 1
    if n_converted <= 3:
        print(month, round(temp_c, 2))       # step 2 check: first three records
    counts[month] = counts.get(month, 0) + 1                  # step 3 (pattern from §2)
    totals_c[month] = totals_c.get(month, 0) + temp_c         # step 4: running total

print(n_converted, n_skipped, n_converted + n_skipped)        # the last number must be 365

means_c = {}                                 # step 4: total -> mean, second loop
for month in totals_c:
    means_c[month] = totals_c[month] / counts[month]
for month, m in means_c.items():
    print(month, round(m, 2))
```

**Step 5 — warmest month** (running max, `None` start because a dict has no `[0]`):

```python
warmest_month = None
warmest_mean = None
for month, m in means_c.items():
    if warmest_mean is None:
        warmest_month, warmest_mean = month, m
    elif m > warmest_mean:
        warmest_month, warmest_mean = month, m
print(warmest_month, round(warmest_mean, 2))
```

**Step 6 — write the summary file, read it back**

```python
out_path = Path("_files") / "monthly_means.csv"        # build a path with /
out_path.parent.mkdir(exist_ok=True)                   # create _files if missing

with out_path.open("w", encoding="utf-8") as f:        # "w" = write (overwrites)
    f.write("month,n_obs,mean_temp_celsius\n")         # header, with \n
    for month in counts:
        f.write(f"{month},{counts[month]},{means_c[month]:.2f}\n")

with out_path.open("r", encoding="utf-8") as f:        # read it back to check
    print(f.read())
```

**Step 7 (optional) — survive junk lines** (wrong field count or non-numeric temperature):

```python
test_lines = [
    "IIB,01.01.22,30.2",
    "IIB,02.01.22",                 # truncated
    "IIB,03.01.22,notanumber",      # not a number
    "IIB,04.01.22,33.4",
]

kept = 0
skipped = 0
for line in test_lines:
    fields = line.strip().split(",")
    if len(fields) != 3:            # truncated line
        skipped += 1
        continue
    try:
        temp_f = float(fields[2])   # float("") and float("notanumber") both raise ValueError
    except ValueError:
        skipped += 1
        continue
    kept += 1
print(kept, skipped)                # 2 2
# On the real file: loop over lines[1:] instead of test_lines. The try/except also
# catches the empty readings, so the explicit `fields[2] == ""` check becomes optional.
```

---

## 9. Planet mass functions  (Ex 17 Q9–Q11 + bonus)

Needs `planet_masses_1e24kg` from §2 (flat dict `name -> mass`). The `_1e24kg` suffix is the unit: always pass masses in 1e24 kg.

```python
def to_earth_masses(mass_1e24kg):                      # Q9
    return mass_1e24kg / planet_masses_1e24kg["earth"]

print(to_earth_masses(planet_masses_1e24kg["jupiter"]))   # 317.92294807370183

def to_planet_masses(mass_1e24kg, reference="jupiter"):    # Q10: one function, default argument
    return mass_1e24kg / planet_masses_1e24kg[reference]

jupiter = planet_masses_1e24kg["jupiter"]
print(to_planet_masses(jupiter, reference="earth"))       # 317.92294807370183
print(to_planet_masses(jupiter, reference="jupiter"))     # 1.0

def mass_in_two_units(mass_1e24kg, reference):             # Q11: return two values
    in_earth = mass_1e24kg / planet_masses_1e24kg["earth"]
    in_reference = mass_1e24kg / planet_masses_1e24kg[reference]
    return in_earth, in_reference

in_earth, in_mars = mass_in_two_units(jupiter, "mars")
print(in_earth, in_mars)
```
````
